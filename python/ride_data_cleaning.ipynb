##### Treating Null Value

```python
# Since,end_lat and end_lang has just 5535 null values in 5.5 million rows , it would be better to drop these rows having null
df=df.dropna(subset=['end_lat','end_lng'])
```

```python
#now check
df.isnull().sum()
```

```python
#start_station_name, start_station_id,end_station_name and end_sation_id have more than 1 million null values 
# this numbers are significient ,it would be better to replace with 'unknown'.
df['start_station_name']=df['start_station_name'].fillna('unknown')
df['start_station_id']=df['start_station_id'].fillna('unknown')
df['end_station_name']=df['end_station_name'].fillna('unknown')
df['end_station_id']=df['end_station_id'].fillna('unknown')
```

```python
# As started_at and ended_at in string datatype ,have to change into date time formate
# Convert to datetime
df['started_at'] = pd.to_datetime(df['started_at']) # pd.to_datetime(df['started_at'],formate='%d-%m-%Y %H:%M:%S'
df['ended_at']   = pd.to_datetime(df['ended_at'])
```
